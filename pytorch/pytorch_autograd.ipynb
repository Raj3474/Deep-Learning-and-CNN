```python
import torch
```

```python
# requires_grad=True: to track gradients for a tensor
x = torch.tensor(3.0, requires_grad=True)
print(x)
```

```python
# f(x) = x ** 2

y = x ** 2
print(y)
```

```python
# .backward(): primary mthod to initiate the
# backward pass, to calculate: dy/dx
y.backward()

# to view the gradient values
x.grad
```

```python
# y = x ^ 2
# z = sin(y)


x = torch.tensor(4.0, requires_grad=True)
y = x ** 2
z = torch.sin(y)

z.backward()
x.grad
```

```python
z
```

```python
# clearing grad
x.grad.zero_() # clear grad before running a new pass
```

```python
# disable gradient tracking, after our training,
# we need to only the forward pass and not backward
# pass to save memory and computation


# option 1 - requires_grad_(False)
# option 2 - detach()
# option 3 - torch.no_grad()
```

```python
x.requires_grad_(False)
```

```python
z = x.detach()
```

```python
z
```

```python
with torch.no_grad():
  y = x ** 2
```
