# Mathematical Operations

## 1. Scalar Operation

```python
x = torch.rand(2, 3)
x
```

```python
# addition
x + 2

# sustraction
x - 2

# multiplication
x * 3

# division
x / 3

# int division
(x * 100) // 3

# mod
((x * 100)//4) % 3
```

```python
torch.clamp(x*10, min=2, max=4) # equivalent of tf.clip_by_value()
```

```python
# torch.sum(), torch.mean(), torch.median(), torch.min() and torch.max()
# product(), std(),, var(), argmax(), argmin(), argsort(),

print('x: ', x)
torch.sum(x, dim=0)
```

```python
torch.argsort(x, dim=1)
```

# 6. Special Functions

```python
k = torch.randint(size=(2, 3), low=0, high=10, dtype=torch.float32)
```

```python
print(k)

torch.log(k)
```

```python
# ReLU

torch.relu(k)
```

```python
# Sigmoid

torch.sigmoid(k)
```

```python
# softmax
torch.softmax(k, dim=1)
```

## Reshapping tensors

```python
a = torch.rand((2, 4, 3))
```

```python
a
```

```python
# Transpose: swaps exactly two specified dimensions
a.transpose(1, 0)
```

```python
a.shape
```

```python
# permute: reorders all dimentions of a tensor
a.permute(2, 0, 1)
```

```python
a.flatten()
```

```python
# squeeze and unsqueeze
a = torch.rand((2, 1, 3))
```

```python
a.shape
```

```python
a.squeeze().shape
```

```python
a.unsqueeze(0).shape
```
